### Store the datasets in database tables

To analyze the data using SQL, it first needs to be loaded into SQLite DB.

Load the `pandas` and `sqlite3` libraries and establish a connection to `FinalDB.db`

### Problem 1

##### Find the total number of crimes recorded in the CRIME table.


```python
table = pd.read_sql("select name from sqlite_master where type = 'table';", con)
print(table)
```

```python
crime_data = pd.read_sql('select * from CHICAGO_CRIME_DATA;', con)
print(crime_data)
```

```python
crimes = pd.read_sql("select count(*) as Number_of_crimes from CHICAGO_CRIME_DATA;", con)
print(crimes)
```

### Problem 2

##### List community area names and numbers with per capita income less than 11000.


```python
census_data = pd.read_sql("select * from CENSUS_DATA;", con)
print(census_data)
```

```python
per_capita_income = pd.read_sql("select COMMUNITY_AREA_NAME, COMMUNITY_AREA_NUMBER from CENSUS_DATA where PER_CAPITA_INCOME < 11000;", con)
print(per_capita_income)
```

### Problem 3

##### List all case numbers for crimes involving minors?(children are not considered minors for the purposes of crime analysis) 


```python
minor_cases = pd.read_sql("select CASE_NUMBER from CHICAGO_CRIME_DATA where DESCRIPTION like '%MINOR%';", con)
print(minor_cases)

```

### Problem 4

##### List all kidnapping crimes involving a child?


```python
kidnapping_crimes = pd.read_sql("select * from CHICAGO_CRIME_DATA where DESCRIPTION like '%CHILD%';", con)
print(kidnapping_crimes)
```

### Problem 5

##### List the kind of crimes that were recorded at schools. (No repetitions)


```python
school_crimes = pd.read_sql("select distinct PRIMARY_TYPE from CHICAGO_CRIME_DATA where LOCATION_DESCRIPTION like '%SCHOOL%';", con)
print(school_crimes)
```

### Problem 6

##### List the type of schools along with the average safety score for each type.


```python
school_data = pd.read_sql("select * from CHICAGO_PUBLIC_SCHOOLS_DATA;", con)
print(school_data)
```

```python
type_of_school = pd.read_sql("""
    SELECT "Elementary, Middle, or High School", AVG(Safety_Score)
    FROM CHICAGO_PUBLIC_SCHOOLS_DATA
    GROUP BY "Elementary, Middle, or High School";
""", con)

print(type_of_school)
```

### Problem 7

##### List 5 community areas with highest % of households below poverty line


```python
community_area_names = pd.read_sql("select COMMUNITY_AREA_NAME, PERCENT_HOUSEHOLDS_BELOW_POVERTY from CENSUS_DATA order by PERCENT_HOUSEHOLDS_BELOW_POVERTY desc limit 5;", con)
print(community_area_names)
```

### Problem 8

##### Which community area is most crime prone? Display the coumminty area number only.


```python
community_crime = pd.read_sql("select COMMUNITY_AREA_NUMBER from CHICAGO_CRIME_DATA group by COMMUNITY_AREA_NUMBER order by count(CASE_NUMBER) desc limit 1; ", con)
print(community_crime)
```

### Problem 9

##### Use a sub-query to find the name of the community area with highest hardship index


```python
name = pd.read_sql("select COMMUNITY_AREA_NAME from CENSUS_DATA where HARDSHIP_INDEX = (select max(HARDSHIP_INDEX) from CENSUS_DATA);", con)
print(name)
```

### Problem 10

##### Use a sub-query to determine the Community Area Name with most number of crimes?


```python
name = pd.read_sql("""
    SELECT COMMUNITY_AREA_NAME
    FROM CENSUS_DATA
    WHERE COMMUNITY_AREA_NUMBER = (
        SELECT COMMUNITY_AREA_NUMBER
        FROM CHICAGO_CRIME_DATA
        GROUP BY COMMUNITY_AREA_NUMBER
        ORDER BY COUNT(CASE_NUMBER) DESC
        LIMIT 1
    );
""", con)

print(name)
```

##### Join version

```python
name= pd.read_sql("""
    SELECT d.COMMUNITY_AREA_NAME
    FROM CENSUS_DATA d
    JOIN CHICAGO_CRIME_DATA c
        ON d.COMMUNITY_AREA_NUMBER = c.COMMUNITY_AREA_NUMBER
    GROUP BY d.COMMUNITY_AREA_NAME
    ORDER BY COUNT(c.CASE_NUMBER) DESC
    LIMIT 1;
""", con)
print(name)    
```
